# 1. Encoder Model

```python
import torch
from transformers import AlbertTokenizer, AlbertForSequenceClassification

repo_id = 'textattack/albert-base-v2-imdb'
tokenizer = AlbertTokenizer.from_pretrained(repo_id)
model = AlbertForSequenceClassification.from_pretrained(repo_id)
```

```python
msg = 'Hello, My dog is cute.'

#긍정인지 부정인지 알아보자

#토크나이저가 의미있는 단어로 분리하고 동시에 index(어휘집 숫자)로 리턴

inputs = tokenizer(msg, return_tensors='pt') # pt = pytorch
```

```python
inputs.keys()
```

```python
with torch.inference_mode():
    #모델 예측
    pred = model(**inputs) #딕셔너리이니까 키워드 가변파라미터로 넘김
```

```python
# 모델 예측값
logits = pred.logits
logits
```

```python
#모델이 예측한 값
pred_class_id = logits.argmax().item()
pred_class_id
```

```python
model.config.id2label[pred_class_id] #LABEL_1 : True를 뜻함.
```
