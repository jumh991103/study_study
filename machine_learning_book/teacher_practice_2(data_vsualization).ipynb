```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import koreanize_matplotlib
```

```python
df = sns.load_dataset('titanic')
df.shape
```

```python
df.columns
```

```python
#문제
# 10대, 20대, 30대 생존률 구하기
#hint
# age 10보다 크고 40보다 작은
# map():  10대, 20대, 30대 class 데이터를 생성
# groupby():  생성한 class 데이터를 통해 그룹핑
# mean(): 그룹핑된 class(10/20/30)의 생존률
df = df[(df['age'] > 9) & (df['age'] < 40)]

def age_class(age):
    if age >29:
        return '30대'   
    elif age > 19:
        return '20대'
    return '10대'

df['age_class'] = df['age'].map(age_class) #얘의 결과는 age에서 10,20,30대 별로 매핑되어서 age_class라는 새로운 컬럼을 만들고 각 로우가 10,20,30대 중에서 뭔지 인자로 들어가있음
df.groupby('age_class')['survived'].agg(['mean','max']) #얘의 결과는 ageclass를 class별로 묶어서 survived의 평균을 냈음. 

#클로드 돌려본결과 그냥 애초에 정확히 알지 못하고 형식만 맞춰서 문제를 풀려해서 틀렸음.
#정확히 공부해서 다시풀어야할듯. 이후의 숙제로 남겨두고 파이썬 끝나는대로 얼른 공부해서 풀어보자 숙제로 하자.
# ->9/12일 문제 풀이했음.
```

## 3. subplots

```python
fig,ax = plt.subplots(2,2,figsize=[15,5]) # (row=2,column=2) 즉 총 그래프 4개를 그리겠다는 뜻

fig.suptitle('도화지 제목') #도화지에 대한 타이틀 : suptitle()

#첫번째 로우 & 왼쪽
ax[0,0].plot(np.arange(5))

#첫번째 로우 & 오른쪽
ax[0,1].set_title('그래프 제목') #그래프에 대한 타이틀 : set_title()
ax[0,1].plot(np.arange(5))
ax[0,1].plot(np.arange(2,7))

#두번째 로우 & 왼쪽
ax[1,0].set_xlabel('x축 이름')
ax[1,0].set_ylabel('y축 이름')

x = range(1,100)
y = np.exp(x)
ax[1,0].plot(x,y)

#두번째 로우 & 오른쪽
x = range(1,10^100)
y = np.log(x)
ax[1,1].plot(x,y)

plt.show()
```

## 5. 그래프 스타일

### 5-3. color

# Matplotlib 심화

## 1. 선그래프

```python
x = np.arange(0,4,0.5)

plt.plot(x, x+1, 'bo')
plt.plot(x,x**2, 'g--')
plt.plot(x, -2*x+3, 'r:')

plt.show()
```

###  1-1. 수평선 그리기 

```python
#axhline(y,xmin, xmax) - xmin,xmax값은 0~1. 0이 왼쪽 끝, 1이 오른쪽 끝 -> x축 범위에 상관없이 0과 1이 끝이다.

plt.axhline(4.0,0.1,0.9, color = 'red', linestyle='--', linewidth = 2)

#hlines(y,xmin,xmax) - (xmin,y)에서 (xmax,y)를 따라 수평선을 표시 -> 좌표 값을 기준으로 끝이 정해진다

plt.hlines(3,1.0, 2.5, color='gray', linestyle='solid', linewidth=3)

plt.axhline(0.0, color = 'yellow', linewidth = 4)

plt.show()
```

### 1-2. 수직선 그리기

```python
#axviline(x,ymin,ymax)
plt.axvline(1.0,0.1,0.8, color= 'yellow', linestyle='--', linewidth=1)


#vlines(x,ymin,ymax)
plt.vlines(3.0, 1.0, 4.0, color = 'red')
plt.show()
```
