```python
# 문제 
# 모든 과목에서 80점 이상인 학생들의 이름을 리스트로 반환하는 함수

# def find_excellent_students(students:list, threshold:int=80) -> list:
#     """
#     4. 모든 과목에서 80점 이상인 학생들의 이름을 리스트로 반환하는 함수
#    
#     Args:
#         students (list): 학생 정보가 담긴 리스트
#         threshold (int): 기준 점수 (기본값: 80)
#    
#     Returns:
#         list: 모든 과목에서 기준점수 이상인 학생들의 이름 리스트
#     """
```

```python
students = [
    {"name": "김철수", "subjects": {"수학": 85, "영어": 90, "과학": 78}},
    {"name": "이영희", "subjects": {"수학": 92, "영어": 88, "과학": 95}},
    {"name": "박민수", "subjects": {"수학": 78, "영어": 85, "과학": 82}},
    {"name": "정지혜", "subjects": {"수학": 88, "영어": 92, "과학": 89}}
]

def find_excellent_students(students:list, threshold:int=80) -> list:
    sub = list(students[0]["subjects"].keys()) #수학, 영어, 과학을 빼내서 리스트로 만듬  = ["수학","영어","과학"]

    up_name = [] #모든과목이 80점 이상인 사람 이름 담을 리스트

    for i in range(len(students)): # 한명씩 검사하기
        for a in range(len(sub)): # 그 한명의 과목의 점수들 검사하기
            if students[i]["subjects"][sub[a]] < threshold: #만약 세개중에 하나 과목이라도 80점 아래면 for문에 break걸리면서 다음사람으로 넘어감
                break
        else: #if문에 안걸리고 for문이 끝까지 돌았다면, 세과목 모두 80점 이상임. for문이 break없이 정상적으로 종료됐을떄 else블록 실행됨.
            up_name.append(students[i]["name"]) #80점 이상인 사람의 이름을 up_name에 추가
    return up_name                               


find_excellent_students(students)


#.keys() - 
```

```python
#claude - 필용없는 변수 줄여줌

students = [
    {"name": "김철수", "subjects": {"수학": 85, "영어": 90, "과학": 78}},
    {"name": "이영희", "subjects": {"수학": 92, "영어": 88, "과학": 95}},
    {"name": "박민수", "subjects": {"수학": 78, "영어": 85, "과학": 82}},
    {"name": "정지혜", "subjects": {"수학": 88, "영어": 92, "과학": 89}}
]

def find_excellent_students(students: list, threshold: int = 80) -> list:
    sub = list(students[0]["subjects"].keys())
    up_name = []
    for student in students:
        for key in sub:
            if student["subjects"][key] < threshold:
                break
        else:
            up_name.append(student["name"])
    return up_name

find_excellent_students(students)
```
