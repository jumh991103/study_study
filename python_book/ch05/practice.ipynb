```python
#팩토리얼

def factorial(n):

    output = 1

    for i in range(1,n+1):
        output *= i
    return output

factorial(3)
```

```python
#팩토리얼 재귀함수로 구현

def factorial_up(n):

    if n==0:
        return 1

    else:
        return n * factorial_up(n-1)

factorial_up(5)
```

```python
#피보나치 수열 

def fibonacci(n):
    if n== 1:
        return 1
    if n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

fibonacci(7)
```

```python
def flatten(data):
    output=[]
    for i in data:

        if type(i) == list:
            output.extend(flatten(i))
        else:
            output.append(i)
    return output

flatten([1,2,3,[4,[5,6]],[7,8],])
```

```python
min = 2
max = 10
total =100
memo = {}

def problem(left, min):
    key = str([left,min])

    if key in memo:
        return memo[key]
    
    if left < 0:
        return 0

    if left == 0:
        return 1

    count = 0
    for i in range(min, max + 1):
        count += problem(left - i, i)

    memo[key] = count

    return count

problem(100, 2)
```

```python
numbers  = list(range(1,10+1))

print("홀수만 추출하기")
print(list(filter(lambda numbers: numbers%2 !=0, numbers)))

print("3이상, 7 미만 추출하기")
print(list(filter(lambda numbers: 3<=numbers<7,numbers)))

print("제곱해서 50미만 추출하기")
print(list(filter(lambda numbers:numbers**2 <=50, numbers)))
```

```python
#도전문제 1
def hanoi(circle, a,b,c):
    if circle == 1:
        print("a","->","b")
    else:
        hanoi(circle-1,a,c,b)
        print("a","->",)
        hanoi(circle-1,c,b,a)
```
