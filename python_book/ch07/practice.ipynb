```python
word = "Hello world"
```

```python
def countLetters(word):
    counter = {}
    for letter in word:
        if letter not in counter:
            counter[letter] = 0 
        counter[letter] += 1
    return counter

countLetters(word)        
```

```python
#만약 setdefault 함수를 쓴다면 
from collections import defaultdict

def countLetters(word):
    counter = {}
    for letter in word:
        counter.setdefault(letter, 0)
        counter[letter] += 1
    return counter

countLetters(word)   
```

```python
tup_list = [
    ("경제","10시 경제기사"),
    ("정치","10시 정치기사"),
    ("사회","10시 사회기사"),
    ("경제","14시 경제기사"),
    ("정치","15시 정치기사"),
]

news_dict = {}  
for k,v in tup_list:
    if news_dict.get(k) is None:
        news_dict[k] = []
    news_dict[k].append(v)
news_dict
```

```python
#defaultdict 사용

def counterLetters(word):
    counter = defaultdict(int) #-> 디폴트값을 int로 할경우 그 키값이 없을떄 0을 추가한다.
    for letter in word:
        counter[letter] +=1 
    return counter

counterLetters(word)
```
