```python
nucleos = input("염기 서열을 입력해주세요")
counter = {
    'a': 0,
    't': 0,
    'g': 0,
    'c': 0
}

for nucleo in nucleos:
    if nucleo in counter:
        counter[nucleo] +=1

for key in counter:
    print(f"{key}의 개수 = {counter[key]}")

```

```python
nucleos =input("염기 서열을 입력해주세요")
count={}

for i in range(0,len(nucleos),3):
    if nucleos[i:i+3] not in count:
        count[nucleos[i:i+3]] = 0
    count[nucleos[i:i+3]] += 1

for i in count:
    print(f"{i}는 {count[i]}")
```
