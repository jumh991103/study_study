```python
# "안녕","안녕하세요"입력 ->"안녕하세요"출력
#"지금 몇 시야?","지금 몇 시예요?" - >지금은 ""시입니다.
import datetime


now = datetime.datetime.now()

a= input("안녕이나 안녕하세요를 입력하세요 : ")
b= input("지금 몇시야 또는 지금 몇시에요라고 물어보세요")

if "안녕" in a:
    print("안녕하세요")


if "몇시" in b:
    print(f"지금 {now.hour}시 입니다.")
```

```python
#정수를 입력받고 2,3,4,5로 나누어떨어지는지 확인하시오
a = int(input("정수를 입력하세요"))

if a % 2 ==0:
    print(f"{a}는 2로 나누어 떨어지는 숫자입니다.")
else : 
    print(f"{a}는 2로 나누어 떨어지는 숫자가 아닙니다.")

if a % 3 ==0:
    print(f"{a}는 3로 나누어 떨어지는 숫자입니다.")
else : 
    print(f"{a}는 3로 나누어 떨어지는 숫자가 아닙니다.")

if a % 4 ==0:
    print(f"{a}는 4로 나누어 떨어지는 숫자입니다.")
else : 
    print(f"{a}는 4로 나누어 떨어지는 숫자가 아닙니다.")

if a % 5 ==0:
    print(f"{a}는 5로 나누어 떨어지는 숫자입니다.")
else : 
    print(f"{a}는 5로 나누어 떨어지는 숫자가 아닙니다.")
```
